# 05 – Results Comparison

This notebook aggregates the evaluation metrics produced in **04_MODEL_EVALUATION** and displays them side‑by‑side with helpful visualisations.

```

import json, pathlib, pandas as pd, matplotlib.pyplot as plt

RESULTS_DIR = pathlib.Path(f'experiment_with_{N_CLASSES}_classes/results')
metrics_file = RESULTS_DIR / 'test_metrics.json'
if not metrics_file.exists():
    raise FileNotFoundError('Run 04_MODEL_EVALUATION.ipynb first.')

with open(metrics_file) as f:
    METRICS = json.load(f)

df = pd.DataFrame.from_dict(METRICS, orient='index')
display(df[['accuracy', 'f1_macro', 'f1_weighted']])
```

```

# Bar plot of macro‑F1 scores
fig, ax = plt.subplots(figsize=(8,4))
df.sort_values('f1_macro').plot(kind='barh', y='f1_macro', ax=ax)
ax.set_xlabel('Macro‑averaged F1')
ax.set_title('Model comparison')
fig.tight_layout()
fig_path = RESULTS_DIR / 'compare_f1.png'
fig.savefig(fig_path, dpi=150)
```
